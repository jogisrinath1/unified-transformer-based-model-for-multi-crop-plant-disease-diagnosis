```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 1️⃣ Accuracy vs Inference Time (Scatter Plot)
# ============================
models = ["Hybrid EfficientNetB0-ViT", "VIT", "ResNet50", "EfficientNetB0"]
accuracy = [99.73, 92.93, 93.99, 67.18]  # in %
inference_time = [28, 35, 42, 58]  # in ms/frame

plt.figure(figsize=(8, 6))
plt.scatter(inference_time, accuracy, color='green', s=100)

for i, model in enumerate(models):
    plt.text(inference_time[i] + 0.5, accuracy[i] + 0.3, model, fontsize=10)

plt.title("Accuracy vs Inference Time", fontsize=16)
plt.xlabel("Inference Time (ms/frame)", fontsize=12)
plt.ylabel("Accuracy (%)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.savefig("accuracy_vs_inference.png", dpi=300)
plt.show()
```

```python
# 2️⃣ Precision & Recall (Bar Chart)
# ============================
models_bar = ["Vision Transformer", "ResNet-50", "EfficientNet", "EfficientNet + ViT"]
precision = [97, 95, 94, 92]
recall = [97, 94, 93, 91]

x = np.arange(len(models_bar))
width = 0.35

plt.figure(figsize=(8, 6))
plt.bar(x - width/2, precision, width, label="Precision", color="orange")
plt.bar(x + width/2, recall, width, label="Recall", color="skyblue")

plt.ylabel("Performance (%)")
plt.title("Precision and Recall Performance")
plt.xticks(x, models_bar, rotation=20)
plt.ylim(85, 100)
plt.legend()
plt.tight_layout()
plt.savefig("precision_recall.png", dpi=300)
plt.show()
```

```python
# 3️⃣ F1-Score Comparison (Bar Chart)
# ============================
f1_scores = [98, 96, 95, 93]

plt.figure(figsize=(8, 6))
plt.bar(models_bar, f1_scores, color=["blue", "green", "orange", "red"])

plt.ylabel("F1-Score (%)")
plt.title("F1-Score Comparison of Models")
plt.ylim(85, 100)
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig("f1_score.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ======================
# Your Accuracy Data
# ======================
models = ["Vision Transformer", "ResNet-50", "EfficientNet", "EfficientNet + ViT (Hybrid)"]
accuracy_1 = [92.93, 93.99, 67.18, 99.73]  # Only Accuracy 1

# ======================
# Plot Single Accuracy Chart
# ======================
x = np.arange(len(models))  # x-axis positions

plt.figure(figsize=(10, 6))
bars = plt.bar(x, accuracy_1, color="#2ecc71", width=0.5)

# Add value labels on top of each bar
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 1, f"{yval:.2f}%",
             ha='center', va='bottom', fontsize=10, fontweight='bold')

# Chart customization
plt.xticks(x, models, rotation=15)
plt.ylabel("Accuracy (%)")
plt.title("Model Accuracy (Accuracy 1)")
plt.ylim(60, 105)
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.tight_layout()
plt.savefig("accuracy_chart.png", dpi=300)
plt.show()

print("✅ Single Accuracy Chart saved as accuracy_chart.png")
```
